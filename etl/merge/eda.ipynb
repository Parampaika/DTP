```python
from pathlib import Path
from io import StringIO
import pandas as pd
import re

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 200)
```

```python
def find_project_root(start: Path = Path.cwd(), marker: str = "configs") -> Path:
    start = start.resolve()
    for parent in [start] + list(start.parents):
        if (parent / marker).exists():
            return parent
    raise FileNotFoundError(f"Не найден корень проекта: папка '{marker}' не найдена")


def df_info_to_string(df: pd.DataFrame) -> str:
    buffer = StringIO()
    df.info(buf=buffer)
    return buffer.getvalue()


def print_section(title: str):
    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)

def normalize_text(s):
    if not isinstance(s, str):
        return ""
    s = s.lower().replace("ё", "е")
    s = re.sub(r"[^а-яa-z0-9\s]", "", s)
    s = re.sub(r"\s+", "_", s)
    return s.strip("_")
```

# Данные

```python
ROOT = find_project_root()

dtp_path = ROOT / "data" / "processed" / "dtp_russia.parquet"
econ_path = ROOT / "data" / "processed" / "econom.parquet"

print_section("1. ПРОВЕРКА ПУТЕЙ")
print(f"ROOT: {ROOT}")
print(f"DTP path: {dtp_path}")
print(f"DTP exists: {dtp_path.exists()}")
print(f"ECON path: {econ_path}")
print(f"ECON exists: {econ_path.exists()}")

if not dtp_path.exists():
    raise FileNotFoundError(f"DTP dataset не найден: {dtp_path}")
if not econ_path.exists():
    raise FileNotFoundError(f"ECON dataset не найден: {econ_path}")

print_section("2. ЗАГРУЗКА ФАЙЛОВ")
df_dtp = pd.read_parquet(dtp_path)
df_econ = pd.read_parquet(econ_path)

print(f"DTP shape: {df_dtp.shape}")
print(f"ECON shape: {df_econ.shape}")
```

## Мёрдж брутфорсом

### Сначала просто банальное совпадение

```python
def clean_for_search(text):
    t = str(text).lower().replace('ё', 'е')
    for w in ['город-герой', 'городской округ', 'город', 'г.']:
        t = t.replace(w, '')
    return re.sub(r'[^а-яa-z0-9]', '', t)

econ_munis = df_econ['municipality'].unique()
dtp_regions = df_dtp['region_id'].dropna().unique()

dtp_search_dict = {
    orig: re.sub(r'[^а-яa-z0-9]', '', str(orig).lower().replace('ё', 'е')) 
    for orig in dtp_regions
}

# Корзины для результатов
single_matches = {}
zero_matches = []
multiple_matches = {}

# 3. Полный перебор
for orig_muni in econ_munis:
    search_term = clean_for_search(orig_muni)
    if len(search_term) < 3:
        continue
        
    matches = []
    for orig_dtp, search_dtp in dtp_search_dict.items():
        if search_term in search_dtp:
            matches.append(orig_dtp)
            
    # Сортируем по корзинам
    if len(matches) == 1:
        single_matches[orig_muni] = matches[0]
    elif len(matches) == 0:
        zero_matches.append((orig_muni, search_term))
    else:
        multiple_matches[orig_muni] = matches

print(f"\n🟢 ТОЧНО 1 СОВПАДЕНИЕ (ГОТОВЫЙ СЛОВАРЬ): {len(single_matches)} шт.")
print("-" * 80)
print("PERFECT_MATCHES = {")
for econ_val, dtp_val in single_matches.items():
    print(f"    '{econ_val}': '{dtp_val}',")
print("}")


print(f"\n🔴 ВООБЩЕ НЕТ СОВПАДЕНИЙ: {len(zero_matches)} шт.")
print("-" * 80)
for orig, search in zero_matches:
    print(f"ECON: '{orig}' (искали по корню: '{search}')")


print(f"\n🟡 МНОЖЕСТВЕННЫЕ СОВПАДЕНИЯ (>1 варианта): {len(multiple_matches)} шт.")
print("-" * 80)
for muni, matches in multiple_matches.items():
    print(f"\nECON: '{muni}' (Найдено вариантов: {len(matches)})")
    sorted_matches = sorted(matches)
    for i in range(0, len(sorted_matches), 3):
        print(" | ".join(sorted_matches[i:i+3]))
```

### Более точечный поиск

```python
print("="*80)
print("АВТОМАТИЧЕСКОЕ РАЗРЕШЕНИЕ МНОЖЕСТВЕННЫХ СОВПАДЕНИЙ")
print("="*80)

def get_city_root(text):
    t = str(text).lower().replace('ё', 'е')
    for w in ['город-герой', 'городской округ', 'город', 'г.']:
        t = t.replace(w, '')
    return re.sub(r'[^а-яa-z0-9]', '', t)

resolved_perfectly = {}
needs_manual_review = {}

for orig_muni, candidates in multiple_matches.items():
    city_root = get_city_root(orig_muni)
    found_exact = None
    
    for cand in candidates:
        if "__" in cand:
            muni_part = cand.split("__", 1)[1]
            muni_part_clean = muni_part.replace('_', '')
            
            if muni_part_clean == city_root:
                found_exact = cand
                break
                
    if found_exact:
        resolved_perfectly[orig_muni] = found_exact
    else:
        needs_manual_review[orig_muni] = candidates

print(f"\n🟢 СПАСЕНО АВТОМАТИЧЕСКИ (нашли точную правую часть): {len(resolved_perfectly)} шт.")
print("-" * 80)
print("RESOLVED_MATCHES = {")
for econ_val, dtp_val in resolved_perfectly.items():
    print(f"    '{econ_val}': '{dtp_val}',")
print("}")

print(f"\n🔴 ОТЛОЖЕНО ПОКА ДУМАЕМ (не нашли точную правую часть): {len(needs_manual_review)} шт.")
print("-" * 80)
for muni, cands in needs_manual_review.items():
    print(f"ECON: '{muni}' (кандидатов: {len(cands)})")
```

```python
print("="*80)
print("ГЛУБОКИЙ ПОИСК ПО КОРНЯМ ДЛЯ ПРОБЛЕМНЫХ ГОРОДОВ")
print("="*80)

broad_search_dict = {
    'Город Барнаул': 'барна',
    'Люберцы': 'любер',
    'Одинцово': 'одинц',
    'Город Нижний Новгород': 'новгород', # поищем просто новгород
    'Город Ростов-на-Дону': 'ростов',
    'Город Балаково': 'балак'
}

dtp_regions = df_dtp['region_id'].dropna().unique()

for econ_muni, stem in broad_search_dict.items():
    print(f"\nИщем ECON: '{econ_muni}' (по корню: '{stem}')")
    
    found_candidates = []
    for dtp_reg in dtp_regions:
        if stem in str(dtp_reg).lower():
            found_candidates.append(dtp_reg)
    else:
        for cand in sorted(found_candidates):
            print(f"   Найдено в ДТП: {cand}")
```

```python
print("="*80)
print("РЕНТГЕН ДЛЯ ОТЛОЖЕННОЙ ПЯТЕРКИ")
print("="*80)

problem_five = ['Город Белгород', 'Город Москва', 'Красногорск', 'Город Дзержинск', 'Город Санкт-Петербург']

for muni in problem_five:
    print(f"\nECON: '{muni}'")
    if muni in multiple_matches:
        cands = multiple_matches[muni]
        
        right_parts = set()
        for c in cands:
                right_parts.add(c) # если вдруг нет подчеркивания
                
        print(f"Что есть в правых частях ДТП ({len(right_parts)} шт):")
        sorted_parts = sorted(list(right_parts))
        for i in range(0, len(sorted_parts), 5):
            print(" | ".join(sorted_parts[i:i+5]))
```

```python
def assign_econ_muni(dtp_id):
    dtp_id = str(dtp_id)
    
    # 1. Ищем в словаре
    if dtp_id in dtp_to_econ:
        return dtp_to_econ[dtp_id]
        
    # 2. Обрабатываем Москву и Питер (любой их район получает экономику всего города)
    if dtp_id.startswith('москва__'):
        return 'Город Москва'
    if dtp_id.startswith('санктпетербург__'):
        return 'Город Санкт-Петербург'
        
    return None

# 3. Применяем и фильтруем по географии
df_dtp['econ_join_key'] = df_dtp['region_id'].apply(assign_econ_muni)

initial_len = len(df_dtp)
df_dtp_geo = df_dtp.dropna(subset=['econ_join_key']).copy()
geo_len = len(df_dtp_geo)

print(f"Всего аварий в ДТП: {initial_len}")
print(f"Аварий в 97 нужных городах (вкл. Мск и СПб): {geo_len}")
print(f"Отброшено аварий (другие регионы/деревни): {initial_len - geo_len}")
```

```python
# 4. Фильтруем по годам и отрезаем лишние текстовые колонки от экономики
common_years = set(df_dtp['year']).intersection(set(df_econ['year']))
df_dtp_ready = df_dtp_geo[df_dtp_geo['year'].isin(common_years)].copy()

# Оставляем в экономике ТОЛЬКО ключи слияния и полезные цифры
econ_features_to_keep = [
    'municipality', 'year', 'Population', 'GAP', 'Avg_Salary', 
    'Employment', 'Street_Length', 'Land_Area', 
    'RTA_dead', 'RTA_serious', 'RTA_minor'
]
df_econ_ready = df_econ[df_econ['year'].isin(common_years)][econ_features_to_keep].copy()

final_len = len(df_dtp_ready)
print(f"\nИз них попало в общие годы {sorted(list(common_years))}: {final_len}")
print(f"Отброшено свежих аварий (2024+): {geo_len - final_len}")

# 5. Итоговый МЕРЖ
df_merged = pd.merge(
    df_dtp_ready,
    df_econ_ready,
    left_on=['econ_join_key', 'year'],
    right_on=['municipality', 'year'],
    how='inner'
)

# Удаляем временные ключи слияния, чтобы датасет был идеально чистым
df_merged.drop(columns=['econ_join_key', 'municipality'], inplace=True)
```

```python
# 1. Убедимся, что у ДТП есть ключ Экономики (если вдруг ячейка перезапускалась)
def assign_econ_muni(dtp_id):
    dtp_id = str(dtp_id)
    if dtp_id in dtp_to_econ:
        return dtp_to_econ[dtp_id]
    if dtp_id.startswith('москва__'):
        return 'Город Москва'
    if dtp_id.startswith('санктпетербург__'):
        return 'Город Санкт-Петербург'
    return None

df_dtp['econ_join_key'] = df_dtp['region_id'].apply(assign_econ_muni)

# 2. АГРЕГАЦИЯ ДТП (Группируем по городу и году)
print("Считаем агрегаты по городам...")
dtp_agg = df_dtp.dropna(subset=['econ_join_key']).groupby(['econ_join_key', 'year']).agg(
    total_dtp=('target', 'count'),
    # Считаем сумму тех аварий, где target 1 (тяжелая) или 2 (смертельная)
    severe_dtp=('target', lambda x: (x >= 1).sum()) 
).reset_index()

# 3. МЕРЖ (Приклеиваем агрегаты к df_econ)
print("Присоединяем статистику к Экономике...")
df_econ_macro = df_econ.merge(
    dtp_agg,
    left_on=['municipality', 'year'],
    right_on=['econ_join_key', 'year'],
    how='left'  # Используем left, чтобы сохранить экономику 2014 года (ДТП там будут NaN)
)

# 4. РАСЧЕТ МЕТРИК
print("Вычисляем метрики безопасности...")
# Доля тяжелых и смертельных аварий (от 0 до 1.0)
df_econ_macro['dtp_severe_share'] = df_econ_macro['severe_dtp'] / df_econ_macro['total_dtp']

# ДТП на 10 000 человек (Population указано в тысячах, поэтому делим на 10)
pop_in_10k = df_econ_macro['Population'] / 10
df_econ_macro['dtp_per_10k'] = df_econ_macro['total_dtp'] / pop_in_10k

# Тяжелые ДТП на 10 000 человек (для более глубокого анализа)
df_econ_macro['severe_per_10k'] = df_econ_macro['severe_dtp'] / pop_in_10k

# Удаляем технический ключ
df_econ_macro.drop(columns=['econ_join_key'], inplace=True, errors='ignore')

# 5. ВЫВОД РЕЗУЛЬТАТОВ
print("\n" + "="*80)
print("ГОТОВО! ПРОВЕРКА РЕЗУЛЬТАТА:")
print("="*80)
print(f"Размер макро-датасета: {df_econ_macro.shape}")
print(f"Пустых значений total_dtp: {df_econ_macro['total_dtp'].isna().sum()} (скорее всего это 2014 год + отброшенные города)")

# Выведем случайные 5 городов за 2021 год, чтобы оценить адекватность цифр
show_cols = ['municipality', 'year', 'Population', 'total_dtp', 'severe_dtp', 'dtp_severe_share', 'dtp_per_10k']
sample_df = df_econ_macro[df_econ_macro['year'] == 2021].dropna(subset=['total_dtp']).sample(5)

print("\nПРИМЕР (5 случайных городов за 2021 год):")
print(sample_df[show_cols].round(3).to_string(index=False))
```
